```python
text = input("Write a text here: ")

words = text.split()

print("Text length:", len(words))

words_over4 = 0
words_upper = 0

for word in words:
    if len(word) > 4:
        words_over4 += 1
    if word[0].isupper():
        words_upper += 1

print("Words greater than 4:", words_over4)
print("Words starting with capital letter:", words_upper)
```
