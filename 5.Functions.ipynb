```python
def count_words(text):
    words = text.lower().split()
    counts = {}

    for word in words:
        if word in counts:
            counts[word] += 1
        else:
            counts[word] = 1
    return counts
```

```python
count_words("HI, hi! Hi hI")
```

```python
def count_words(text):
    words = text.lower().split()
    counts = {}
    do_not_count = ",?!@#."
    for word in words:
        word = word.strip(do_not_count)
        
        if word in counts:
            counts[word] += 1
        else:
            counts[word] = 1
    return counts
```

```python
count_words("HI, hi! Hi hI")
```

```python
def top_word(text):
    words = text.lower().split()
    count = {}
    
    for word in words:
        word = word.strip(".,!?")
        if word in count:
            count[word] += 1
        else:
            count[word] = 1

    if not count:
            return None,0

    max_word = None
    max_count = 0

    for word in count:
        if count[word] > max_count:
            max_count = count[word]
            max_word = word
        

    return max_word,count[max_word]


        
```
