```python
mylist = [4,9,1,7,3]

print("Length of list:", len(mylist))
print("First element:", mylist[0])
print("Last element:", mylist[-1])
print("First two element:", mylist[0:2])
print("Last two element:", mylist[-1:-3:-1]);

abovefive_count = 0
even_count = 0

for num in mylist:
    if num > 5:
        abovefive_count += 1
    if num % 2 == 0:
        even_count += 1
print("Greater than five:", abovefive_count)
print("Even:", even_count);

mylist.append(12)
mylist.pop(0)
mylist.pop()
print(mylist)
```
